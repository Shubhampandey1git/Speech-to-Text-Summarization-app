## **IMPORTS**

```python
import pandas as pd
import re
import os
```

**Load Data**

```python
HINDI_CSV = "../data/processed/asr/hindi/train.csv"

df = pd.read_csv(HINDI_CSV)

print("Total samples:", len(df))
df.head()
```

**English presence**

```python
def has_english(text):
    return bool(re.search(r"[A-Za-z]", str(text)))

df["has_english"] = df["text"].apply(has_english)
```

```python
total = len(df)
english_count = df["has_english"].sum()
pure_hindi_count = total - english_count

print("Total samples:", total)
print("Contains English:", english_count)
print("Pure Hindi:", pure_hindi_count)

print("\nPercentage with English:", round(english_count / total * 100, 2), "%")
```

**Look at the examples**

```python
print("=== Samples WITH English ===")
df[df["has_english"] == True]["text"].head(10).tolist()
```

```python
print("=== Pure Hindi samples ===")
df[df["has_english"] == False]["text"].head(10).tolist()
```

## **Filtering**

```python
HINDI_CSV = "../data/processed/asr/hindi/train.csv"
OUTPUT_CSV = "../data/processed/asr/hindi/clean/train.csv"

df = pd.read_csv(HINDI_CSV)

# detect English
def has_english(text):
    return bool(re.search(r"[A-Za-z]", str(text)))

df["has_english"] = df["text"].apply(has_english)

# filter out rows with English
clean_df = df[df["has_english"] == False].copy()

# drop helper column
clean_df = clean_df.drop(columns=["has_english"])

print("Original:", len(df))
print("Cleaned :", len(clean_df))
print("Removed :", len(df) - len(clean_df))
```

***save***

```python
os.makedirs("../data/processed/asr/hindi/clean", exist_ok=True)

clean_df.to_csv(OUTPUT_CSV, index=False)

print("Saved at:", OUTPUT_CSV)
```
